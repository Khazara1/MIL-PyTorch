```python
from metrics import BinaryMetricsCalculator, MulticlassMetricsCalculator
import torch
import matplotlib.pyplot as plt
import numpy as np
```

```python
preds = torch.load("outputs.pth")
targets = torch.load("targets.pth")
```

```python
metrics_calculator = BinaryMetricsCalculator()
```

```python
test_accuracy, test_f1_score, test_auprc, test_auroc, test_precision, test_recall, conf_matrix = metrics_calculator.calculate(preds, targets)
```

```python
thresholds = np.arange(0.01, 1, 0.01)
```

```python
for thresh in thresholds:
    metrics_calculator = BinaryMetricsCalculator(threshold=thresh)
    test_accuracy, test_f1_score, test_auprc, test_auroc, test_precision, test_recall, conf_matrix = metrics_calculator.calculate(preds, targets)
    tn, fp = conf_matrix[0]
    fn, tp = conf_matrix[1]
    specificity = tn / (tn + fp)

    print(f"Threshold: {thresh:.4f} | Prec: {test_precision:.4f} | Recall: {test_recall:.4f} | Specificity: {specificity:.4f} | F1: {test_f1_score:.4f}")
```

```python
print("test_accuracy:", test_accuracy)
print("test_f1_score:", test_f1_score)
print("test_auprc:", test_auprc)
print("test_auroc:", test_auroc)
print("test_precision:", test_precision)
print("test_recall:", test_recall)
print("conf_matrix:", conf_matrix)
```
